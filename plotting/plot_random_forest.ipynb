```python
import json
import os

import pandas as pd

# Plot the accuracy

dir = '../data/results/experiments/random_forest_experiment_5'

data_df = pd.read_csv(os.path.join(dir, 'data.csv')).drop(columns=['Unnamed: 0'])

with open(os.path.join(dir, 'config.json')) as f:
    config = json.load(f)

data_df
```

```python
res = data_df.groupby('n_clusters')[['accuracy', 'recall', 'f1_score', 'mean_cv_score']].mean().round(3)
```
